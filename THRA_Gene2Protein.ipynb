## **Step 3: Load Annotation Database**

The AnnotationHub database provides access to Ensembl genomic annotations. The AH119325 database contains human genome annotations.

```python
!R -e "library(ensembldb); library(AnnotationHub); ah <- AnnotationHub(); print(ah); edb <- ah[['AH119325']]; print(edb)"
```

## **Step 4: Extract Gene Information**

The genes() function is used to retrieve information related to the THRA gene.

```python
!R -e "library(ensembldb); library(AnnotationHub); ah <- AnnotationHub(); edb <- ah[['AH119325']]; gene_info <- genes(edb, filter = ~ gene_id == 'ENSG00000126351'); print(gene_info)"
```

### Observations:

*   Gene Name: THRA
*   Ensembl Gene ID: ENSG00000126351
*   Chromosome: 17
*   Gene Type: protein_coding
*   Canonical Transcript: ENST00000450525

## **Step 5: Canonical Transcript: ENST00000450525**

The transcripts() function retrieves all transcript isoforms associated with the THRA gene.

```python
!R -e "library(ensembldb); library(AnnotationHub); ah <- AnnotationHub(); edb <- ah[['AH119325']]; tx <- transcripts(edb, filter = ~ gene_id == 'ENSG00000126351'); print(tx)"
```

### Observations:

*   Multiple transcript isoforms were identified.
*   Most transcripts are protein_coding.
*   One transcript belongs to the retained_intron biotype.
*   The presence of multiple transcripts demonstrates alternative splicing.


## **Step 6: Extract Protein Information**

The proteins() function retrieves protein isoforms corresponding to THRA transcripts.

```python
!R -e "library(ensembldb); library(AnnotationHub); ah <- AnnotationHub(); edb <- ah[['AH119325']]; prot <- proteins(edb, filter = ~ gene_id == 'ENSG00000126351'); print(prot)"
```

### Observations:

*   Multiple ENSP protein IDs were identified.
*   Protein sequences corresponding to transcript isoforms were extracted.
*   Different transcripts produce different protein isoforms.

## **Step 7: Save Transcript Information**

The transcript information extracted from the THRA gene was saved into a TSV file for downstream analysis and record keeping.

```python
!R -e "library(ensembldb); library(AnnotationHub); ah <- AnnotationHub(); edb <- ah[['AH119325']]; tx <- transcripts(edb, filter = ~ gene_id == 'ENSG00000126351'); write.table(as.data.frame(tx), file='THRA_transcripts.tsv', sep='\t', row.names=FALSE, quote=FALSE)"
```

## **Step 8: Save Protein Information**

Protein isoform information associated with THRA transcripts was saved into a TSV file.

```python
!R -e "library(ensembldb); library(AnnotationHub); ah <- AnnotationHub(); edb <- ah[['AH119325']]; prot <- proteins(edb, filter = ~ gene_id == 'ENSG00000126351'); write.table(as.data.frame(prot), file='THRA_proteins.tsv', sep='\t', row.names=FALSE, quote=FALSE)"
```

## **Step 9: Save Gene Information**

Gene-level information related to THRA was exported into a TSV file.

```python
!R -e "library(ensembldb); library(AnnotationHub); ah <- AnnotationHub(); edb <- ah[['AH119325']]; gene_info <- genes(edb, filter = ~ gene_id == 'ENSG00000126351'); write.table(as.data.frame(gene_info), file='THRA_gene_info.tsv', sep='\t', row.names=FALSE, quote=FALSE)"
```

## **Step 10: Count Transcript and Protein Isoforms**

The total number of transcript and protein isoforms associated with THRA was calculated.

```python
!R -e "library(ensembldb); library(AnnotationHub); ah <- AnnotationHub(); edb <- ah[['AH119325']]; tx <- transcripts(edb, filter = ~ gene_id == 'ENSG00000126351'); prot <- proteins(edb, filter = ~ gene_id == 'ENSG00000126351'); cat('Total transcripts:', nrow(as.data.frame(tx)), '\n'); cat('Total proteins:', nrow(as.data.frame(prot)), '\n')"
```

### Observation

THRA contains multiple transcript and protein isoforms, demonstrating transcript diversity and alternative splicing.

## **Step 11: Transcript–Protein Mapping**

A mapping table was generated to associate transcript IDs (ENST) with their corresponding protein IDs (ENSP).

```python
!R -e "library(ensembldb); library(AnnotationHub); ah <- AnnotationHub(); edb <- ah[['AH119325']]; prot <- proteins(edb, filter = ~ gene_id == 'ENSG00000126351'); mapping <- data.frame(transcript_id = prot$tx_id, protein_id = prot$protein_id); print(mapping); write.table(mapping, file='THRA_tx_protein_mapping.tsv', sep='\t', row.names=FALSE, quote=FALSE)"
```

### Observation

Each transcript isoform corresponds to a specific protein isoform, demonstrating the Gene → Transcript → Protein relationship.

## **Step 12: Canonical Transcript Identification**

The canonical transcript of THRA was identified using the tx_is_canonical field.

```python
!R -e "library(ensembldb); library(AnnotationHub); ah <- AnnotationHub(); edb <- ah[['AH119325']]; tx <- as.data.frame(transcripts(edb, filter = ~ gene_id == 'ENSG00000126351')); print(tx[, c('tx_id', 'tx_is_canonical')])"
```

### Observation

The canonical transcript identified for THRA is:

- ENST00000450525
